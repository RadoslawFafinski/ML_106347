# 1. Import bibliotek i wczytanie danych

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

np.set_printoptions(precision=4, suppress=True)

iris = load_iris()
X = iris.data
y = iris.target

print("Kształt danych:", X.shape)
print("Nazwy cech:", iris.feature_names)
print("Gatunki:", iris.target_names)
```

# 2. Centrowanie danych

```python
mean_vector = np.mean(X, axis=0)
X_centered = X - mean_vector

print("Średnie cech:")
print(mean_vector)

print("\nŚrednie po centrowaniu (powinny być bliskie zeru):")
print(np.mean(X_centered, axis=0))
```
